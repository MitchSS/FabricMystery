# Rebind Semantic Model

Updates the Direct Lake connection for AetherSM to point at the AetherLH Lakehouse SQL endpoint
in the target workspace. Run this after deployment to bind the semantic model to the correct data source.

```python
%pip install semantic-link-labs
```

```python
from sempy_labs import directlake

dataset = "AetherSM"  # Semantic model name
workspace = "{{WORKSPACE_NAME}}"  # Target workspace name — injected at deploy time
source = "AetherLH"  # Lakehouse name
source_type = "Lakehouse"
source_workspace = "{{WORKSPACE_NAME}}"  # Same workspace

directlake.update_direct_lake_model_connection(
    dataset=dataset,
    workspace=workspace,
    source=source,
    source_type=source_type,
    source_workspace=source_workspace,
)

print(f"Rebound {dataset} to {source} in workspace '{workspace}'")
```
